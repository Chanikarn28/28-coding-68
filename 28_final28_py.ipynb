```python
import ipywidgets as widgets
from IPython.display import display, clear_output

def check_budget_ipywidgets(b):
    with output_widget:
        clear_output()
        try:
            budget_val = entry_budget.value
            spent_val = entry_spent.value

            if spent_val < budget_val:
                print(f"💰 ใช้น้อยกว่างบ\nเหลือ {budget_val - spent_val:.2f} บาท 🎉")
            elif spent_val == budget_val:
                print("😄 ใช้พอดีกับงบเป๊ะเลย!")
            else:
                print(f"⚠️ ใช้เกินงบ\nเกิน {spent_val - budget_val:.2f} บาท 😢")
        except ValueError:
            print("Error: กรุณาใส่ตัวเลขเท่านั้นนะคะ 💕")

# Input widgets
entry_budget = widgets.FloatText(
    description='งบประมาณ (บาท):',
    value=0.0,
    min=0,
    step=1.0
)

entry_spent = widgets.FloatText(
    description='ใช้ไปแล้ว (บาท):',
    value=0.0,
    min=0,
    step=1.0
)

# Button widget
check_button = widgets.Button(
    description='✨ ตรวจสอบ ✨',
    button_style='info' # 'success', 'info', 'warning', 'danger' or ''
)
check_button.on_click(check_budget_ipywidgets)

# Output widget to display results
output_widget = widgets.Output()

# Define a light pink color
light_pink = '#FFD1DC' # Or any other light pink hex code

# Display the widgets with light pink background
display(
    widgets.VBox([
        widgets.Label("🌷 โปรแกรมตรวจสอบการใช้เงิน 🌷", style={'font_weight': 'bold', 'font_size': '16px', 'background_color': light_pink}),
        entry_budget,
        entry_spent,
        check_button,
        output_widget
    ], layout=widgets.Layout(background_color=light_pink))
)
```

```python
#เวอร์ชั่นใช้พี่เอไอช่วยเหลือ
def calculate_water_bill():
    try:
        # รับค่าจำนวนหน่วยน้ำจากผู้ใช้
        units = float(input("กรุณาใส่จำนวนหน่วยน้ำที่ใช้: "))

        if units <= 0:
            print("จำนวนหน่วยต้องมากกว่า 0")
            return

        bill = 0

        if units <= 20:
            # กรณีใช้ไม่เกิน 20 หน่วย
            bill = units * 10
        else:
            # กรณีใช้เกิน 20 หน่วย
            # 20 หน่วยแรก (20 * 10) + หน่วยที่เหลือ (units - 20) * 15
            bill = (20 * 10) + ((units - 20) * 15)

        print(f"--- สรุปค่าใช้จ่าย ---")
        print(f"จำนวนหน่วยที่ใช้: {units} หน่วย")
        print(f"รวมค่าน้ำทั้งสิ้น: {bill:,.2f} บาท")

    except ValueError:
        print("กรุณากรอกเฉพาะตัวเลขเท่านั้น")

# เรียกใช้งานฟังก์ชัน
calculate_water_bill()
```
